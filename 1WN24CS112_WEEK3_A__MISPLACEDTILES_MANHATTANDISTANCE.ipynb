```python
from heapq import heappush, heappop


goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)



def misplaced_tiles(state):

    count = 0

    for i in range(9):
        if state[i] != 0 and state[i] != goal[i]:
            count += 1

    return count


def manhattan_distance(state):

    distance = 0

    for i in range(9):

        if state[i] == 0:
            continue

        current_row = i // 3
        current_col = i % 3

        goal_index = state[i] - 1
        goal_row = goal_index // 3
        goal_col = goal_index % 3

        distance += abs(current_row - goal_row)
        distance += abs(current_col - goal_col)

    return distance


def get_neighbors(state):

    neighbors = []

    zero_position = state.index(0)

    row = zero_position // 3
    col = zero_position % 3


    moves = [
        (-1, 0),
        (1, 0),
        (0, -1),
        (0, 1)
    ]

    for dr, dc in moves:

        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_position = new_row * 3 + new_col

            new_state = list(state)


            new_state[zero_position], new_state[new_position] = \
                new_state[new_position], new_state[zero_position]

            neighbors.append(tuple(new_state))

    return neighbors


def print_state(state):

    for i in range(0, 9, 3):
        print(state[i], state[i + 1], state[i + 2])

    print()



def a_star(start, heuristic):

    open_list = []

    g = 0


    h = heuristic(start)


    f = g + h

    heappush(open_list, (f, g, start, [start]))

    visited = set()

    while open_list:

        f, g, current, path = heappop(open_list)


        if current in visited:
            continue

        visited.add(current)

        if current == goal:

            print("\nGoal reached!")
            print("Number of moves:", g)
            print("Solution path:\n")

            for state in path:
                print_state(state)

            return


        for neighbor in get_neighbors(current):

            if neighbor not in visited:

                new_g = g + 1
                new_h = heuristic(neighbor)
                new_f = new_g + new_h

                new_path = path + [neighbor]

                heappush(
                    open_list,
                    (new_f, new_g, neighbor, new_path)
                )

    print("No solution found.")




print("8-Puzzle A* Search")
print("------------------")

print("\nEnter the initial state:")
print("Enter 0 for blank space.")

start = []

for i in range(9):
    value = int(input("Enter tile " + str(i + 1) + ": "))
    start.append(value)

start = tuple(start)

print("\nChoose heuristic:")
print("1. Misplaced Tiles")
print("2. Manhattan Distance")

choice = int(input("Enter your choice: "))

if choice == 1:

    print("\nUsing Misplaced Tiles heuristic...")
    a_star(start, misplaced_tiles)

elif choice == 2:

    print("\nUsing Manhattan Distance heuristic...")
    a_star(start, manhattan_distance)

else:

    print("Invalid choice.")
```
