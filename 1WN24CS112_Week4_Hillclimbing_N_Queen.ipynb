```python

def cost_function(board):
    n = len(board)
    attacking_pairs = 0


    for i in range(n):
        for j in range(i + 1, n):


            if board[i] == board[j]:
                attacking_pairs += 1

            elif abs(board[i] - board[j]) == abs(i - j):
                attacking_pairs += 1

    return attacking_pairs


def hill_climbing(board):
    n = len(board)
    current_cost = cost_function(board)

    while current_cost != 0:

        best_board = board.copy()
        best_cost = current_cost


        for i in range(n):
            for j in range(i + 1, n):

                neighbor = board.copy()

                neighbor[i], neighbor[j] = neighbor[j], neighbor[i]

                neighbor_cost = cost_function(neighbor)

                if neighbor_cost < best_cost:
                    best_cost = neighbor_cost
                    best_board = neighbor.copy()


        if best_cost >= current_cost:
            print("Local minimum reached!")
            print("Current board:", board)
            print("Current cost:", current_cost)
            return None


        board = best_board
        current_cost = best_cost

        print("Current board:", board)
        print("Current cost:", current_cost)


    return board


n = int(input("Enter the number of queens: "))


board = []

print("Enter the row position for each column (0 to", n - 1, "):")

for i in range(n):
    row = int(input("Column " + str(i) + ": "))
    board.append(row)

print("\nInitial board:", board)
print("Initial cost:", cost_function(board))

solution = hill_climbing(board)

if solution is not None:
    print("\nSolution found!")
    print("Final board:", solution)
    print("Final cost:", cost_function(solution))
else:
    print("\nNo solution found from this initial state.")

```
